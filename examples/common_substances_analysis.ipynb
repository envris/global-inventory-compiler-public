# Finding substances present on multiple inventories
This notebook demonstrates how to query a compiled global inventory to identify substances that occur across multiple regulatory inventories. Such queries are useful for highlighting widely used or widely regulated substances.

**Prerequisites**
- Download a copy of the [example global inventory](https://github.com/envris/global-inventory-compiler/tree/main/data) from the GIC repository.

## Step 1: Load the inventory and inspect the available list tags
Create an Inventory instance and print the list tags present in the dataset.

```python
from gic import Inventory

# Load the precompiled inventory.
inv = Inventory("input/global_inventory.jsonl")

print(inv.get_list_data().values())
```

## Step 2: Define the inventory tags of interest
Select the major inventory list tags to include in the cross-inventory comparison. Many other list tags represent specific regulatory measures and can be excluded for this purpose.

```python
inventories = [
    "AU-AIIC",  # Australia
    "CA-DSL",  # Canada
    "CL-HSIUN",  # Chile
    "CN-IECSC",  # China
    "EU-REACH_PR",  # Europe
    "EU-REACH_R",  # Europe
    "JP-CSCL",  # Japan
    "KR-KECL",  # South Korea
    "NZ-NZIoC",  # New Zealand
    "TW-CSI",  # Taiwan
    "US-TSCA",  # United States
]
```

## Step 3: Run the query
The *get_by_list_tags* method returns identifiers of the requested type for substances that satisfy the specified list-based criteria. Setting *shared=True* restricts the result to substances that appear in every selected inventory.

```python
common_sids = inv.get_by_list_tags("sid", inventories, shared=True)

print(f"Number of substances in inventory: {len(inv.get_sids())}")
print(f"Number of substances common to these inventories: {len(common_sids)}")
```

## Step 4: Export the results (optional)
SIDs are assigned during compilation and have limited meaning outside the context of a specific inventory. Exporting the matching records makes them easier to review in tabular form.

```python
results = inv.export_as_data_frame(common_sids)

# Save the results for inspection.
results.to_csv("output/common_substances.csv")
results.to_xlsx("output/common_substances.xlsx")
```
