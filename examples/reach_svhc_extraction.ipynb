# Extracting identifiers from the EU REACH Candidate List of Substances of Very High Concern (SVHC)
This notebook demonstrates how to extract chemical names and key identifiers from the EU REACH SVHC list and save them as a JSONL file suitable for use in GIC workflows and global inventory compilation.

**Prerequisites**
- Export the list as a CSV file from the [ECHA website](https://echa.europa.eu/candidate-list-table).

## Step 1: Import and initialise the extractor
Use the generic CSV extractor provided by the GIC.

```python
from gic import GenericCsvExtractor

# Initialise the generic CSV extractor.
extractor = GenericCsvExtractor()
```

## Step 2: Configure the extractor
Set the input and output paths and map the relevant CSV columns to GIC fields.

```python
# Input and output files.
extractor.input_file = "input/candidate-list-of-svhc-for-authorisation-export.csv"
extractor.output_file = "output/reach.jsonl"
extractor.errors_file = "output/reach_errors.tsv"

# Column-to-field mapping: chemical name and EC number.
extractor.columns = {
    1: "chemical_name",
    3: "ec_number",
}

# Assign the source list tag to each extracted row.
extractor.constants = {"list_tag": ["EU-REACH_SVHC"]}

# File layout settings.
extractor.start_row = 14
extractor.delimiter = "\t"
```

## Step 3: Run the extraction and inspect the output
Read the source file, write the extracted records to *reach.jsonl*, and log validation issues to *reach_errors.tsv*.

```python
df = extractor.extract()

# Inspect example records as dictionaries.
print(df[10].as_dict())
print(df[433].as_dict())
```
