# Extracting identifiers from the Australian Inventory of Industrial Chemicals (AIIC)
This notebook demonstrates how to extract chemical names and key identifiers from the AIIC and save the result as a JSONL file suitable for use in a global inventory compilation.

**Prerequisites**
- Download the Excel version of the AIIC from the [AICIS website](https://services.industrialchemicals.gov.au/search-inventory/).

## Step 1: Import and initialise the extractor
Use the generic XLSX extractor provided by the GIC.

```python
from gic import GenericXlsxExtractor

# Initialise the generic XLSX extractor.
extractor = GenericXlsxExtractor()
```

## Step 2: Configure the extractor
Set the input and output paths and map the source columns to GIC fields. The constants attribute assigns the AIIC list tag to every extracted row.

Note that generic_id_<1-5> fields can be used for identifiers that are not explicitly supported by the GIC. These fields receive only generic validation for null-like values, unsafe text, and non-specific placeholders.

```python
# Downloaded AIIC source file.
extractor.input_file = (
    "input/Full list of chemicals on the Inventory - 9 January 2026 -V8.xlsx"
)

# Output files.
extractor.output_file = "output/aiic.jsonl"
extractor.errors_file = "output/aiic_errors.tsv"

# Column-to-field mapping: CR number and chemical name.
extractor.columns = {
    1: "generic_id_2",
    3: "chemical_name",
}

# Assign the source list tag to each extracted row.
extractor.constants = {"list_tag": ["AU-AIIC"]}

# Skip the first two header rows.
extractor.start_row = 3
```

## Step 3: Run the extraction and inspect the output
Read the Excel file, write the extracted records to *aiic.jsonl*, and log any validation issues to *aiic_errors.tsv*.

```python
df = extractor.extract()

# Inspect example records as dictionaries.
print(df[10].as_dict())
print(df[30283].as_dict())
```
